```python
# ========================
# 3. PROFESSIONAL INPUT SYSTEM
# ========================
class InputSystem:
    def __init__(self):
        self.create_widgets()
        self.setup_ui()
        # Initialize dropdowns with default values
        self.course.value = 'B.Tech'
        self._update_course_selection({'new': self.course.value})

    def create_widgets(self):
        # Header with requested changes
        self.header = widgets.HTML("""
        <div style="background:#2c3e50; color:white; padding:20px; border-radius:5px;">
            <h1 style="margin:0; text-align:center; font-weight:bold; text-transform:uppercase;">GRAPHIC ERA HILL UNIVERSITY BHIMTAL</h1>
            <p style="text-align:center; margin:5px 0 0; font-weight:bold; text-transform:uppercase;">UNIVERSITY TIMETABLE GENERATOR</p>
        </div>
        """)

        # Enhanced course options
        self.course = widgets.Dropdown(
            options=['B.Tech', 'B.Pharma', 'B.Sc. Agriculture', 'BCA', 'BBA', 'B.Com', 'MBA', 'MCA', 'M.Tech'],
            description='Select Course:',
            style={'description_width': '150px'},
            layout=widgets.Layout(width='80%')
        )

        # Year options will be populated dynamically
        self.year = widgets.Dropdown(
            options=[],
            description='Select Year:',
            style={'description_width': '150px'},
            layout=widgets.Layout(width='80%')
        )

        # Subject options will be populated dynamically
        self.subject = widgets.Dropdown(
            options=[],
            description='Select Subject:',
            style={'description_width': '150px'},
            layout=widgets.Layout(width='80%')
        )

        # Enhanced faculty options
        self.faculty = widgets.Dropdown(
            options=list(Config.FACULTIES.keys()),
            description='Select Faculty:',
            style={'description_width': '150px'},
            layout=widgets.Layout(width='80%')
        )

        # Faculty Leave Date
        self.leave_date = widgets.DatePicker(
            description='Leave Date:',
            disabled=False,
            style={'description_width': '150px'},
            layout=widgets.Layout(width='80%')
        )

        # Separate substitute faculty options
        self.substitute = widgets.Dropdown(
            options=list(Config.SUBSTITUTE_FACULTIES.keys()),
            description='Substitute Faculty:',
            style={'description_width': '150px'},
            layout=widgets.Layout(width='80%')
        )

        # Additional Options
        self.include_labs = widgets.Checkbox(
            value=True,
            description='Include Laboratory Sessions',
            disabled=False,
            indent=False
        )

        # Generation mode
        self.generation_mode = widgets.RadioButtons(
            options=['Complete Timetable', 'Subject-wise', 'Automatic'],
            description='Generation Mode:',
            disabled=False,
            layout=widgets.Layout(width='80%')
        )

        # Generate Button
        self.generate_btn = widgets.Button(
            description='Generate Timetable',
            button_style='success',
            icon='calendar',
            layout=widgets.Layout(width='300px', height='50px')
        )

        # Status Output
        self.status = widgets.Output()

    def _update_course_selection(self, change):
        if change['new']:
            course = change['new']
            # Update years based on course duration
            if course in ['B.Tech', 'B.Pharma', 'B.Sc. Agriculture']:
                self.year.options = ['First', 'Second', 'Third (Pre-Final)', 'Fourth (Final)']
                self.year.value = 'First'
            elif course in ['BCA', 'BBA', 'B.Com']:
                self.year.options = ['First', 'Second', 'Third (Final)']
                self.year.value = 'First'
            elif course in ['MBA', 'MCA', 'M.Tech']:
                self.year.options = ['First', 'Second (Final)']
                self.year.value = 'First'

            # Update subjects for the selected year
            self._update_year_selection({'new': self.year.value})

    def _update_year_selection(self, change):
        if self.course.value and change['new']:
            year = change['new'].split(' ')[0]  # Remove (Final...) part
            course = self.course.value
            theory_subjects = Config.COURSES[course]['subjects'][year]['theory']
            lab_subjects = Config.COURSES[course]['subjects'][year]['lab']
            self.subject.options = theory_subjects + lab_subjects
            if theory_subjects:
                self.subject.value = theory_subjects[0]
            elif lab_subjects:
                self.subject.value = lab_subjects[0]

    def setup_ui(self):
        # Set up observers
        self.course.observe(self._update_course_selection, names='value')
        self.year.observe(self._update_year_selection, names='value')

        # Form Sections
        course_section = widgets.VBox([
            widgets.HTML("<h3 style='color:#3498db;'>Course Information</h3>"),
            self.course,
            self.year,
            self.subject,
            self.include_labs,
            self.generation_mode
        ], layout=widgets.Layout(width='45%', margin='0 10px'))

        faculty_section = widgets.VBox([
            widgets.HTML("<h3 style='color:#3498db;'>Faculty Information</h3>"),
            self.faculty,
            self.leave_date,
            self.substitute
        ], layout=widgets.Layout(width='45%', margin='0 10px'))

        # Main form layout
        self.form = widgets.VBox([
            self.header,
            widgets.HBox([course_section, faculty_section]),
            widgets.HBox([
                self.generate_btn
            ], layout=widgets.Layout(justify_content='center')),
            self.status
        ], layout=widgets.Layout(width='90%', margin='20px auto'))

    def display(self):
        display(self.form)
```

```python
class TimetableGenerator:
    def __init__(self):
        self.population_size = 200
        self.generations = 100
        self.mutation_rate = 0.15
        self.elitism = 0.1

    def initialize_population(self, constraints):
        """Create initial population of timetables"""
        population = []
        for _ in range(self.population_size):
            timetable = self.generate_timetable(constraints)
            population.append(timetable)
        return population

    def generate_timetable(self, constraints):
        """Generate complete timetable with labs"""
        timetable = []
        course_info = Config.COURSES[constraints['course']]
        year = constraints['year'].split(' ')[0]  # Remove (Final...) part

        # For automatic mode, generate complete timetable
        if constraints.get('generation_mode', 'Complete Timetable') == 'Automatic':
            # Add all theory classes
            for subject in course_info['subjects'][year]['theory']:
                timetable.append(self.create_class_slot(
                    constraints, subject, 'Lecture'
                ))

            # Add all lab sessions if enabled
            if constraints.get('include_labs', True):
                for lab in course_info['subjects'][year]['lab']:
                    # Labs typically take 2 consecutive slots
                    slot = self.create_class_slot(constraints, lab, 'Lab')
                    timetable.append(slot)

                    # Add consecutive slot for same lab
                    next_slot = slot.copy()
                    next_slot['time'] = self.get_next_time_slot(slot['time'])
                    timetable.append(next_slot)

        # For subject-wise mode, generate only selected subject
        elif constraints.get('generation_mode') == 'Subject-wise':
            subject = constraints.get('subject', '')
            if subject in course_info['subjects'][year]['theory']:
                timetable.append(self.create_class_slot(
                    constraints, subject, 'Lecture'
                ))
            elif subject in course_info['subjects'][year]['lab']:
                # Labs take 2 slots
                slot = self.create_class_slot(constraints, subject, 'Lab')
                timetable.append(slot)
                next_slot = slot.copy()
                next_slot['time'] = self.get_next_time_slot(slot['time'])
                timetable.append(next_slot)

        # For complete timetable mode (default)
        else:
            # Add theory classes
            for subject in course_info['subjects'][year]['theory']:
                timetable.append(self.create_class_slot(
                    constraints, subject, 'Lecture'
                ))

            # Add lab sessions if enabled
            if constraints.get('include_labs', True):
                for lab in course_info['subjects'][year]['lab']:
                    # Labs typically take 2 consecutive slots
                    slot = self.create_class_slot(constraints, lab, 'Lab')
                    timetable.append(slot)

                    # Add consecutive slot for same lab
                    next_slot = slot.copy()
                    next_slot['time'] = self.get_next_time_slot(slot['time'])
                    timetable.append(next_slot)

        return timetable

    def create_class_slot(self, constraints, subject, class_type):
        """Create individual timetable slot"""
        # Check if faculty is on leave and needs substitution
        faculty = constraints.get('faculty', None)
        leave_date = constraints.get('leave_date', None)
        substitute = constraints.get('substitute', None)

        if (faculty and leave_date and
            str(leave_date) in Config.FACULTIES[faculty]['unavailable'] and
            substitute):
            assigned_faculty = substitute
        elif faculty:
            assigned_faculty = faculty
        else:
            assigned_faculty = self.select_faculty(subject, constraints)

        return {
            'course': constraints['course'],
            'year': constraints['year'].split(' ')[0],  # Remove (Final...) part
            'subject': subject,
            'faculty': assigned_faculty,
            'room': self.select_room(class_type),
            'day': random.choice(Config.DAYS),
            'time': random.choice(Config.TIME_SLOTS),
            'type': class_type,
            'duration': 2 if class_type == 'Lab' else 1
        }

    def select_faculty(self, subject, constraints):
        """Select appropriate faculty considering expertise and availability"""
        # First try main faculty
        available_faculty = [
            f for f, data in Config.FACULTIES.items()
            if subject in data['expertise']
        ]

        # If no main faculty available, try substitutes
        if not available_faculty:
            available_faculty = [
                f for f, data in Config.SUBSTITUTE_FACULTIES.items()
                if subject in data['expertise']
            ]

        return random.choice(available_faculty) if available_faculty else "To Be Assigned"

    def select_room(self, class_type):
        """Select appropriate room based on class type"""
        if class_type == 'Lab':
            return random.choice(list(Config.LAB_ROOMS.keys()))
        return random.choice(list(Config.LECTURE_ROOMS.keys()))

    def get_next_time_slot(self, current_time):
        """Get consecutive time slot for labs"""
        times = Config.TIME_SLOTS
        idx = times.index(current_time)
        return times[idx + 1] if idx + 1 < len(times) else current_time

    def calculate_fitness(self, timetable):
        """Evaluate timetable quality with multiple constraints"""
        fitness = 1000  # Starting fitness

        # Track conflicts
        room_slots = defaultdict(set)
        faculty_slots = defaultdict(set)
        student_slots = defaultdict(set)

        for slot in timetable:
            # Room conflict check
            room_key = (slot['room'], slot['day'], slot['time'])
            if room_key in room_slots:
                fitness -= 50  # Heavy penalty for room double-booking
            room_slots[room_key].add(slot['subject'])

            # Faculty conflict check
            faculty_key = (slot['faculty'], slot['day'], slot['time'])
            if faculty_key in faculty_slots:
                fitness -= 40  # Penalty for faculty double-booking
            faculty_slots[faculty_key].add(slot['subject'])

            # Student group conflict (same year/course)
            student_key = (slot['course'], slot['year'], slot['day'], slot['time'])
            if student_key in student_slots:
                fitness -= 30
            student_slots[student_key].add(slot['subject'])

            # Room type suitability
            if slot['type'] == 'Lab' and 'Lab' not in Config.ROOMS[slot['room']]['type']:
                fitness -= 30

            # Faculty-subject match
            if slot['faculty'] != "To Be Assigned":
                if (slot['faculty'] in Config.FACULTIES and
                    slot['subject'] not in Config.FACULTIES[slot['faculty']]['expertise']):
                    fitness -= 60
                elif (slot['faculty'] in Config.SUBSTITUTE_FACULTIES and
                      slot['subject'] not in Config.SUBSTITUTE_FACULTIES[slot['faculty']]['expertise']):
                    fitness -= 60

            # Room capacity consideration
            if Config.ROOMS[slot['room']]['capacity'] < 30 and slot['type'] == 'Lecture':
                fitness -= 20

        return fitness

    def crossover(self, parent1, parent2):
        """Create child timetable through crossover"""
        child = []
        cutoff = random.randint(1, len(parent1)-1)
        child.extend(parent1[:cutoff])
        child.extend(parent2[cutoff:])
        return child

    def mutate(self, timetable):
        """Apply mutations to timetable"""
        for i in range(len(timetable)):
            if random.random() < self.mutation_rate:
                # Mutate either day, time, room, or faculty
                mutation_type = random.choice(['day', 'time', 'room', 'faculty'])
                if mutation_type == 'day':
                    timetable[i]['day'] = random.choice(Config.DAYS)
                elif mutation_type == 'time':
                    timetable[i]['time'] = random.choice(Config.TIME_SLOTS)
                elif mutation_type == 'room':
                    timetable[i]['room'] = self.select_room(timetable[i]['type'])
                elif mutation_type == 'faculty':
                    timetable[i]['faculty'] = self.select_faculty(timetable[i]['subject'], {})
        return timetable

    def evolve(self, constraints):
        """Run full genetic algorithm optimization"""
        population = self.initialize_population(constraints)

        for generation in range(self.generations):
            # Evaluate fitness
            population = sorted(population, key=lambda x: self.calculate_fitness(x), reverse=True)

            # Elitism: keep best individuals
            elites = population[:int(self.elitism * self.population_size)]

            # Create new generation
            children = []
            while len(children) < self.population_size - len(elites):
                parent1, parent2 = random.choices(population[:50], k=2)  # Select from top 50%
                child = self.crossover(parent1, parent2)
                child = self.mutate(child)
                children.append(child)

            population = elites + children

        return max(population, key=lambda x: self.calculate_fitness(x))
```

```python
class TimetableRenderer:
    def __init__(self):
        self.styles = """
        <style>
            body {
                font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif;
                background: linear-gradient(135deg, #f5f7fa 0%, #c3cfe2 100%);
                margin: 0;
                padding: 20px;
            }
            .timetable-container {
                max-width: 1200px;
                margin: 20px auto;
                background: white;
                border-radius: 10px;
                box-shadow: 0 10px 30px rgba(0,0,0,0.1);
                overflow: hidden;
            }
            .timetable-header {
                background: linear-gradient(135deg, #667eea 0%, #764ba2 100%);
                color: white;
                padding: 25px;
                text-align: center;
            }
            .timetable-header h1 {
                margin: 0;
                font-size: 28px;
                font-weight: 700;
                text-transform: uppercase;
            }
            .timetable-header h2 {
                margin: 10px 0 0;
                font-size: 22px;
                font-weight: 600;
                text-transform: uppercase;
            }
            .timetable-header p {
                margin: 5px 0 0;
                font-size: 16px;
            }
            .timetable {
                width: 100%;
                border-collapse: collapse;
                margin: 20px 0;
            }
            .timetable th {
                background: linear-gradient(135deg, #6B73FF 0%, #000DFF 100%);
                color: white;
                padding: 12px;
                text-align: center;
                position: sticky;
                top: 0;
                font-weight: 600;
            }
            .timetable td {
                border: 1px solid #e0e0e0;
                padding: 10px;
                text-align: center;
            }
            .timetable tr:nth-child(even) {
                background-color: #f8f9fa;
            }
            .lecture {
                background-color: #e3f2fd;
            }
            .lab {
                background-color: #ffebee;
            }
            .footer {
                text-align: center;
                padding: 15px;
                background: #f5f5f5;
                color: #666;
                font-size: 0.9em;
            }
            .controls {
                margin: 20px 0;
                display: flex;
                justify-content: center;
                gap: 15px;
            }
            .btn {
                padding: 10px 20px;
                border: none;
                border-radius: 5px;
                cursor: pointer;
                font-weight: bold;
                transition: all 0.3s;
                box-shadow: 0 2px 5px rgba(0,0,0,0.1);
            }
            .btn:hover {
                transform: translateY(-2px);
                box-shadow: 0 4px 8px rgba(0,0,0,0.15);
            }
            .btn-primary {
                background: linear-gradient(135deg, #667eea 0%, #764ba2 100%);
                color: white;
            }
            .btn-success {
                background: linear-gradient(135deg, #2AF598 0%, #009EFD 100%);
                color: white;
            }
            .stats-card {
                background: white;
                margin: 20px;
                padding: 20px;
                border-radius: 8px;
                box-shadow: 0 5px 15px rgba(0,0,0,0.05);
            }
            .stats-card h3 {
                color: #2c3e50;
                margin-top: 0;
                border-bottom: 2px solid #3498db;
                padding-bottom: 5px;
            }
            .stats-grid {
                display: grid;
                grid-template-columns: repeat(3, 1fr);
                gap: 15px;
            }
            .stats-item {
                background: #f8f9fa;
                padding: 10px;
                border-radius: 5px;
            }
            .stats-item h4 {
                color: #3498db;
                margin: 0 0 10px 0;
            }
            .highlight {
                background-color: #fffde7;
                padding: 2px 5px;
                border-radius: 3px;
                font-weight: 500;
            }
        </style>
        """

    def show_loading(self, current_gen, total_gens):
        """Show animated loading screen"""
        return HTML(f"""
        <div style="text-align:center; padding:40px;">
            <div class="spinner-border" style="width:3rem; height:3rem; color:#3498db;"></div>
            <h3 style="margin-top:20px; color:#2c3e50;">Optimizing Timetable</h3>
            <p>Generation: {current_gen}/{total_gens}</p>
            <p style="color:#7f8c8d;">Using genetic algorithm with 200 chromosomes</p>
        </div>
        """)

    def generate_html(self, timetable, constraints):
        """Generate complete HTML output"""
        # Group by day
        day_slots = {day: [] for day in Config.DAYS}
        for slot in timetable:
            day_slots[slot['day']].append(slot)

        # Create HTML structure
        html = [f"""
        <!DOCTYPE html>
        <html lang="en">
        <head>
            <meta charset="UTF-8">
            <meta name="viewport" content="width=device-width, initial-scale=1.0">
            <title>Generated Timetable - {constraints['course']} {constraints['year']}</title>
            {self.styles}
        </head>
        <body>
            <div class="timetable-container">
                <div class="timetable-header">
                    <h1>GRAPHIC ERA HILL UNIVERSITY BHIMTAL</h1>
                    <h2>UNIVERSITY TIMETABLE GENERATOR</h2>
                    <p>{Config.ACADEMIC_YEAR} - {Config.SEMESTER} Semester | Generated on {datetime.now().strftime('%d %B %Y at %H:%M')}</p>
                </div>

                <div class="stats-card">
                    <h3>Timetable Statistics</h3>
                    <div class="stats-grid">
                        <div class="stats-item">
                            <h4>Course Information</h4>
                            <p><strong>Program:</strong> <span class="highlight">{constraints['course']}</span></p>
                            <p><strong>Year:</strong> <span class="highlight">{constraints['year']}</span></p>
                            <p><strong>Labs Included:</strong> <span class="highlight">{'Yes' if constraints.get('include_labs', True) else 'No'}</span></p>
                        </div>
                        <div class="stats-item">
                            <h4>Faculty Information</h4>
                            <p><strong>Primary Faculty:</strong> <span class="highlight">{constraints.get('faculty', 'Auto-assigned')}</span></p>
                            <p><strong>Substitute Faculty:</strong> <span class="highlight">{constraints.get('substitute', 'None')}</span></p>
                            <p><strong>Leave Date:</strong> <span class="highlight">{constraints.get('leave_date', 'None')}</span></p>
                        </div>
                        <div class="stats-item">
                            <h4>Schedule Summary</h4>
                            <p><strong>Lectures:</strong> <span class="highlight">{len([s for s in timetable if s['type']=='Lecture'])}</span></p>
                            <p><strong>Labs:</strong> <span class="highlight">{len([s for s in timetable if s['type']=='Lab'])}</span></p>
                            <p><strong>Total Sessions:</strong> <span class="highlight">{len(timetable)}</span></p>
                        </div>
                    </div>
                </div>
        """]

        # Add day-wise tables
        for day, slots in day_slots.items():
            if slots:  # Only add days with classes
                html.append(f"""
                <h3 style="color:#2c3e50; border-bottom:2px solid #3498db; padding-bottom:5px; margin: 20px 20px 0 20px;">{day}</h3>
                <table class="timetable" style="margin: 0 20px 20px 20px;">
                    <thead>
                        <tr>
                            <th>Time</th>
                            <th>Subject</th>
                            <th>Type</th>
                            <th>Faculty</th>
                            <th>Room</th>
                            <th>Building</th>
                        </tr>
                    </thead>
                    <tbody>
                """)

                for slot in sorted(slots, key=lambda x: x['time']):
                    faculty_name = slot['faculty']
                    if faculty_name in Config.SUBSTITUTE_FACULTIES:
                        faculty_display = f"{faculty_name} (Substitute)"
                    else:
                        faculty_display = faculty_name

                    html.append(f"""
                    <tr class="{slot['type'].lower()}">
                        <td>{slot['time']}</td>
                        <td><strong>{slot['subject']}</strong></td>
                        <td>{slot['type']}</td>
                        <td>{faculty_display}</td>
                        <td>{slot['room']}</td>
                        <td>{Config.ROOMS[slot['room']]['building']}</td>
                    </tr>
                    """)

                html.append("""
                    </tbody>
                </table>
                """)

        # Add footer and controls
        html.append(f"""
            <div class="controls">
                <button class="btn btn-primary" onclick="window.print()">Print Timetable</button>
                <button class="btn btn-success" onclick="location.reload()">Generate New</button>
            </div>

            <div class="footer">
                <p>Generated by AI Timetable Scheduling System</p>
                <p>© {datetime.now().year} Graphic Era Hill University Bhimtal | Department of Computer Science & Engineering</p>
            </div>
        </div>
        </body>
        </html>
        """)

        return ''.join(html)

    def show_in_new_tab(self, timetable, constraints):
        """Open timetable in new browser tab with colorful UI"""
        html_content = self.generate_html(timetable, constraints)

        # Generate random filename
        import uuid
        filename = f"timetable_{uuid.uuid4().hex[:6]}.html"

        # Create and auto-click download link
        display(HTML(f"""
        <script>
            var blob = new Blob([{json.dumps(html_content)}], {{type: "text/html"}});
            var url = URL.createObjectURL(blob);

            // Open in new tab
            var win = window.open(url, '_blank');

            // Focus the new tab
            if (win) {{
                win.focus();
            }} else {{
                // Fallback for popup blockers
                alert('Please allow popups for this site to view the timetable');
                var a = document.createElement('a');
                a.href = url;
                a.download = '{filename}';
                document.body.appendChild(a);
                a.click();
                document.body.removeChild(a);
            }}
        </script>

        <div style="
            background: linear-gradient(135deg, #f5f7fa 0%, #c3cfe2 100%);
            padding: 20px;
            border-radius: 10px;
            text-align: center;
            margin: 20px 0;
            box-shadow: 0 4px 8px rgba(0,0,0,0.1);
        ">
            <h3 style="color: #27ae60; margin-top: 0;">Timetable Generated Successfully!</h3>
            <p style="color: #555; font-size: 16px;">The timetable has been opened in a new tab.</p>
            <p style="color: #777; font-size: 14px;">If the new tab didn't open automatically, please check your browser's pop-up settings.</p>
            <button onclick="window.location.reload()" style="
                background: #3498db;
                color: white;
                border: none;
                padding: 8px 16px;
                border-radius: 4px;
                cursor: pointer;
                margin-top: 10px;
            ">
                Generate Another Timetable
            </button>
        </div>
        """))

    def show_statistics(self, timetable):
        """Display analytical visualizations"""
        df = pd.DataFrame(timetable)

        # Faculty workload
        faculty_load = df['faculty'].value_counts().reset_index()
        faculty_load.columns = ['Faculty', 'Classes']

        # Room utilization
        room_usage = df['room'].value_counts().reset_index()
        room_usage.columns = ['Room', 'Classes']
        room_usage['Type'] = room_usage['Room'].apply(
            lambda x: 'Lab' if x in Config.LAB_ROOMS else 'Lecture'
        )

        # Create plots
        plt.figure(figsize=(15, 10))

        plt.subplot(2, 2, 1)
        sns.barplot(x='Faculty', y='Classes', data=faculty_load)
        plt.title('Faculty Workload Distribution')
        plt.xticks(rotation=45)

        plt.subplot(2, 2, 2)
        sns.barplot(x='Room', y='Classes', hue='Type', data=room_usage)
        plt.title('Room Utilization')
        plt.xticks(rotation=45)

        plt.subplot(2, 1, 2)
        day_dist = df['day'].value_counts()
        plt.pie(day_dist, labels=day_dist.index, autopct='%1.1f%%')
        plt.title('Class Distribution Across Days')

        plt.tight_layout()
        plt.show()
```

```python
class TimetableSchedulerApp:
    def __init__(self):
        self.input_system = InputSystem()
        self.generator = TimetableGenerator()
        self.renderer = TimetableRenderer()
        self.setup_callbacks()

    def setup_callbacks(self):
        def on_generate(btn):
            with self.input_system.status:
                clear_output()

                # Show loading animation
                display(self.renderer.show_loading(0, self.generator.generations))

                # Get constraints from UI
                constraints = {
                    'course': self.input_system.course.value,
                    'year': self.input_system.year.value,
                    'subject': self.input_system.subject.value,
                    'faculty': self.input_system.faculty.value,
                    'leave_date': self.input_system.leave_date.value,
                    'substitute': self.input_system.substitute.value,
                    'include_labs': self.input_system.include_labs.value,
                    'generation_mode': self.input_system.generation_mode.value
                }

                # Simulate generation process with progress
                for i in range(1, self.generator.generations + 1):
                    time.sleep(0.05)
                    hue = min(120, (i/100)*120)  # Calculate hue in Python first
                    display(Javascript(f"""
                        document.querySelector(".spinner-border").style.color =
                            'hsl({hue}, 100%, 50%)';
                        document.querySelector("p").innerText = "Generation: {i}/{self.generator.generations}";
                    """))

                # Generate timetable
                timetable = self.generator.evolve(constraints)

                # Display results
                clear_output()
                self.renderer.show_in_new_tab(timetable, constraints)

                # Show in notebook for reference
                display(HTML("<h3 style='color:#27ae60;'>Timetable Preview</h3>"))
                display(HTML(self.renderer.generate_html(timetable, constraints)))

                # Show analytics
                display(HTML("<h3 style='color:#2c3e50; margin-top:30px;'>Performance Analytics</h3>"))
                self.renderer.show_statistics(timetable)

        self.input_system.generate_btn.on_click(on_generate)

    def run(self):
        self.input_system.display()
```
